<h2>Probability Calculation</h2>

```python
import nltk
from nltk.util import ngrams
from collections import Counter
from functools import reduce
import operator

# -----------------------------
# Corpus
# -----------------------------
sentences = [
    "<s> Quality furniture improves every home </s>".split(),
    "<s> Damaged furniture ruins every home </s>".split()
]

# -----------------------------
# Vocabulary
# -----------------------------
vocab = sorted(set(word for sent in sentences for word in sent))
V = len(vocab)

print("="*60)
print("Vocabulary")
print(vocab)
print("Vocabulary Size =", V)

# ==========================================================
# UNIGRAM
# ==========================================================

print("\n" + "="*60)
print("UNIGRAM")

unigram_counts = Counter()

for sent in sentences:
    unigram_counts.update(sent)

N = sum(unigram_counts.values())

print("\nUnigram Counts")
for word in vocab:
    print(f"{word:12} : {unigram_counts[word]}")

print("\nUnigram Probabilities")
for word in vocab:
    print(f"P({word}) = {unigram_counts[word]/N:.4f}")

# Sentence Probability
def unigram_probability(sentence):
    prob = 1

    # Ignore <s> and </s>
    for word in sentence[1:-1]:
        prob *= unigram_counts[word]/N

    return prob

positive = sentences[0]
negative = sentences[1]

print("\nPositive Sentence Probability")
print(unigram_probability(positive))

print("\nNegative Sentence Probability")
print(unigram_probability(negative))

# ==========================================================
# BIGRAM
# ==========================================================

print("\n" + "="*60)
print("BIGRAM")

bigram_counts = Counter()
history_bigram = Counter()

for sent in sentences:
    bgs = list(ngrams(sent,2))

    bigram_counts.update(bgs)

    for bg in bgs:
        history_bigram[bg[0]] += 1

print("\nBigram Counts")
for bg,count in bigram_counts.items():
    print(bg,"=",count)

def bigram_probability(w1,w2):
    return (bigram_counts[(w1,w2)] + 1)/(history_bigram[w1] + V)

def sentence_bigram_probability(sentence):

    prob = 1

    print()

    for bg in ngrams(sentence,2):

        p = bigram_probability(bg[0],bg[1])

        print(f"P({bg[1]} | {bg[0]}) = {p:.4f}")

        prob *= p

    return prob

print("\nPositive Sentence")
bp = sentence_bigram_probability(positive)
print("\nSentence Probability =",bp)

print("\nNegative Sentence")
bn = sentence_bigram_probability(negative)
print("\nSentence Probability =",bn)

# ==========================================================
# TRIGRAM
# ==========================================================

print("\n" + "="*60)
print("TRIGRAM")

trigram_counts = Counter()
history_trigram = Counter()

for sent in sentences:

    tgs = list(ngrams(sent,3))

    trigram_counts.update(tgs)

    for tg in tgs:

        history_trigram[(tg[0],tg[1])] += 1

print("\nTrigram Counts")

for tg,count in trigram_counts.items():
    print(tg,"=",count)

def trigram_probability(w1,w2,w3):

    history = (w1,w2)

    return (trigram_counts[(w1,w2,w3)] + 1)/(history_trigram[history] + V)

def sentence_trigram_probability(sentence):

    prob = 1

    print()

    for tg in ngrams(sentence,3):

        p = trigram_probability(tg[0],tg[1],tg[2])

        print(f"P({tg[2]} | {tg[0]} {tg[1]}) = {p:.4f}")

        prob *= p

    return prob

print("\nPositive Sentence")
tp = sentence_trigram_probability(positive)
print("\nSentence Probability =",tp)

print("\nNegative Sentence")
tn = sentence_trigram_probability(negative)
print("\nSentence Probability =",tn)

# ==========================================================
# FINAL RESULT
# ==========================================================

print("\n" + "="*60)
print("FINAL PROBABILITY SCORES")
print("="*60)

print(f"\nUnigram Positive : {unigram_probability(positive)}")
print(f"Unigram Negative : {unigram_probability(negative)}")

print(f"\nBigram Positive  : {bp}")
print(f"Bigram Negative  : {bn}")

print(f"\nTrigram Positive : {tp}")
print(f"Trigram Negative : {tn}")
```

<h2>Test Sentence</h2>

```python
# ==========================================================
# TEST SENTENCE
# ==========================================================

test_sentence = "<s> Quality furniture ruins every home </s>".split()

print("\n" + "="*60)
print("TEST SENTENCE")
print(" ".join(test_sentence))

# ==========================================================
# UNIGRAM TEST SENTENCE PROBABILITY
# ==========================================================

print("\nUNIGRAM CALCULATION")

uni_prob = 1

for word in test_sentence[1:-1]:      # Ignore <s> and </s>

    p = unigram_counts[word] / N

    print(f"P({word}) = {p:.4f}")

    uni_prob *= p

print("\nFinal Unigram Probability =", uni_prob)


# ==========================================================
# BIGRAM TEST SENTENCE PROBABILITY
# ==========================================================

print("\n" + "="*60)
print("BIGRAM CALCULATION")

bigram_prob = 1

for bg in ngrams(test_sentence,2):

    p = bigram_probability(bg[0],bg[1])

    print(f"P({bg[1]} | {bg[0]}) = {p:.4f}")

    bigram_prob *= p

print("\nFinal Bigram Probability =", bigram_prob)


# ==========================================================
# TRIGRAM TEST SENTENCE PROBABILITY
# ==========================================================

print("\n" + "="*60)
print("TRIGRAM CALCULATION")

trigram_prob = 1

for tg in ngrams(test_sentence,3):

    p = trigram_probability(tg[0],tg[1],tg[2])

    print(f"P({tg[2]} | {tg[0]} {tg[1]}) = {p:.4f}")

    trigram_prob *= p

print("\nFinal Trigram Probability =", trigram_prob)
```

<h2>Perplexity Score</h2>

```python
import math

# ======================================================
# PERPLEXITY FUNCTION
# ======================================================

def perplexity(probability, N):
    return (1 / probability) ** (1 / N)

# -----------------------------
# Number of predicted words
# -----------------------------

uni_N = len(test_sentence[1:-1])      # 5 words

bigram_N = len(list(ngrams(test_sentence,2)))   # 6 transitions

trigram_N = len(list(ngrams(test_sentence,3)))  # 5 transitions

# -----------------------------
# Calculate Perplexities
# -----------------------------

uni_pp = perplexity(uni_prob, uni_N)
bigram_pp = perplexity(bigram_prob, bigram_N)
trigram_pp = perplexity(trigram_prob, trigram_N)

print("\n" + "="*60)
print("PERPLEXITY SCORES")
print("="*60)

print(f"Unigram Perplexity : {uni_pp:.2f}")
print(f"Bigram Perplexity  : {bigram_pp:.2f}")
print(f"Trigram Perplexity : {trigram_pp:.2f}")
```
